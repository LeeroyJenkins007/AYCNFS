```python
import numpy as np
import matplotlib.pyplot as plt

from astro_tools.data.atmospheres import USSA1976
```

The 1976 U.S. Standard Atmospheric (USSA) is a multi-layered atmosphere model from 0 MSL to ~84km with modifications for above 84km. 

Lets Test out the USSA1976 implementation.

```python
alts = np.arange(0, 80000, 1)
ussa = USSA1976()
temps = [ussa.molecular_scale_temperature(alt) for alt in alts]
pressure = [ussa.pressure(alt) / 1000 for alt in alts]
density = [ussa.density(alt) for alt in alts]
```

Now, that we have the data needed, we can start plotting them

```python
fig, axs = plt.subplots(1, 3, figsize=(11,5), sharey=True)
fig.suptitle("U.S. Standard Atmosphere 1976")

axs[0].set_ylabel("Geometric Altitude [km]")
axs[0].plot(temps, alts / 1000)
axs[0].set_xlabel("Temperature [K]")
axs[0].grid(True)

axs[1].set_ylabel("Geometric Altitude [km]")
axs[1].set_xlabel("Density [kg/m^3]")
axs[1].plot(density, alts / 1000)
axs[1].grid(True)

axs[2].set_ylabel("Geometric Altitude [km]")
axs[2].set_xlabel("Pressure [kPa]")
axs[2].plot(pressure, alts / 1000)
axs[2].grid(True)
```
